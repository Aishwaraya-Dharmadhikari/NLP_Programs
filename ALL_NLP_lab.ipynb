```python
import spacy

nlp=spacy.load("en_core_web_sm")
ner_categories=["PERSON", "ORG", "GPE"]
```

```python
text="""Robert Downey Jr. is an American actor and producer.
He is best known for his roles in films as Iron Man, The Avengers, and Sherlock Homies. Downey has won serveal awards for his acting, including two screen actors guild awards and a Golden F=Globe Award."""

doc=nlp(text)
```

```python
entites= []
for ent in doc.ents:
    if ent.label_ in ner_categories:
        entites.append((ent.text, ent.label_))
```

```python
for entity in entites:
    print(f"{entity[0]}: {entity[1]}")
```

```python
from spacy import displacy
displacy.render(doc, style="ent")
```

```python
import re
import nltk
from nltk.corpus import stopwords
```

```python
nltk.download('stopwords')
```

```python
text = """Dr. A. P. J. Abdul Kalam, the Missile Man of India, inspired millions through his books

and speeches."""
```

```python
text = text.lower()
text = re.sub(r'[^\w\s]', '', text)
text = re.sub(r'\d+', '', text)
```

```python
stop_words = set(stopwords.words('english'))
words = text.split()
filtered_words = [word for word in words if word not in stop_words]
```

```python
print("Processed Text:", ' '.join(filtered_words))
```

```python
from nltk.stem import PorterStemmer
import nltk
nltk.download('punkt')
```

```python
stemmer = PorterStemmer()
text = "Farmers in Maharashtra are cultivating sugarcane and wheat."
words = nltk.word_tokenize(text)
```

```python
stemmed_words = [stemmer.stem(word) for word in words]
```

```python
print("Stemmed Words:", stemmed_words)
```

```python
from nltk.stem import WordNetLemmatizer
import nltk
nltk.download('punkt')
nltk.download('wordnet')
```

```python
lemmatizer = WordNetLemmatizer()
text = "Peacocks are dancing in the gardens during the monsoon."
words = nltk.word_tokenize(text)
```
